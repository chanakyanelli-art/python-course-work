```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print(np.dot(A, B))
```

```python
print(np.linalg.det(A))
```

```python
print(np.linalg.inv(A))
```

```python
eigenvalues, eigenvectors = np.linalg.eig(A)
print(eigenvalues)
```

```python
C = np.array([5, 11])
solution = np.linalg.solve(A, C)
print(solution)
```
