```python
import pandas as pd

prices=[2999, 15999, 52999, 4999, 1999]
products=["Wireless Earbuds", "Smartphone", "Laptop", "Smartwatch", "Bluetooth Speaker"]
product_prices = pd.Series(prices,index=products)
print(product_prices)
```

```python
print("Mean:", product_prices.mean())
```

```python
print("Sum:", product_prices.sum())
```

```python
print("Head (First 3 Element):\n", product_prices.head(3))
```

```python
print("Tail (Last 2 Element):\n", product_prices.tail(2))
```

```python
print("apply (Adding 18% GST):\n", product_prices.apply(lambda x: f'₹{x+(x*0.18)}'))
```

```python
print("Map (Formating as Currency):\n", product_prices.map(lambda x: f"₹{x}.00"))
```

```python
print(product_prices.sort_values())
```

```python
print(product_prices.sort_index())
```

```python
print(product_prices.sort_index(ascending=False))
```

```python
print(product_prices.value_counts())
```

```python
import pandas as pd

data = {
"Products": ["Wireless Earbuds", "Smartphone", "Laptop", "Smartwatch", "Bluetooth Speaker"],
"Brand": ["SundMax", "TechNova", "ByteCore", "TimeTrack", "EchoBoom"],
"Prices": [2999, 15999, 52999, 4999, 1999],
"Stock": [50, 30, 20, 40, 60],
"BestSeller": [True,False,False,True,True]
}
df = pd.DataFrame(data)
print(df)
```

```python
print(df.shape)
print(df.columns)
print(df.info())
```

```python
print(df.head(3))
print(df.tail(2))
```

```python
print(df)
print(df.loc[4,"Products"])
print(df.loc[4,"Products"])
print(df.loc[4,"Stock"])

print(df.iloc[4, 0])
print(df.iloc[2, 1])
print(df.iloc[3, 3])
```

```python
print(df.iloc[3])
print(df.iloc[2])
```

```python
print(df)
print(df.loc[df['BestSeller']==True])
print(df.loc[df['Stock']<40])
print(df.loc[df['Prices']<1000])
```

```python
 df_renamed = df.rename(columns={"Price": "Cost"})
 print(df_renamed)
```

```python
df_grouped = df.groupby("Brand").agg({"Prices": "sum", "Stock": "sum"})
print(df_grouped)
```

```python
grouped=df.groupby("Brand").agg({"Prices": ["mean", "max", "min"]})
print(grouped)
```

```python
grouped = df.groupby("Brand")["Prices"].mean()
print(grouped)
```

```python
import numpy as np

data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [np.nan, 2, 3, np.nan, 5],
    'C': [10, np.nan, np.nan, 40, 40]
}
df = pd.DataFrame(data)
print("Original DataFrame with Missing Values:")
print(df)
```

```python
df_dropna = df.dropna()
print("\nDataFrame after dropping rows with missing values:")
print(df_dropna)
```

```python
df_fillna = df.fillna(0)
print(df_fillna)
```

```python
df_interpolate = df.interpolate()
print("\nDataFrame after filling NaN values using interpolation:")
print(df_interpolate)
```
